```python
# 6. Add a New Column Using Exsisting Columns 
```

```python
usd_rate = {"USD": 1.0, "EUR": 1.08, "GBP": 1.27, "CAD": 0.73, "GHS": 0.085, "NGN": 0.00065}
df["order_total_usd"] = (df["order_total"] * df["currency"].map(usd_rate)).round(2)
df["price_per_item_usd"] = (df["order_total_usd"] / df["items"]).round(2)
df[["order_id", "currency", "order_total", "items", "order_total_usd", "price_per_item_usd"]].head()
```

```python
# 7. Filter the data 
# Filter 1: high value orders 
high_value_df = df[df["order_total_usd"] > 500]
print("Records with order value over 500 USD:", len(high_value_df))
display(high_value_df[["order_id", "country", "product_category", "order_total_usd", "returned"]].head())
```

```python
# Filter 2: Fast Deliveries 
fast_delivery_df = df[df["delivery_days"] <= 3]
print("Records with delivery in 3 days or fewer:", len(fast_delivery_df))
display(fast_delivery_df[["order_id", "country", "delivery_days", "rating"]].head())
```

```python
# 9. Grouping and Aggregation
# Categorical Column
df["order_total_usd"] = pd.to_numeric(df["order_total_usd"], errors='coerce')
df["returned"] = pd.to_numeric(df["returned"], errors='coerce')
df["rating"] = pd.to_numeric(df["rating"], errors='coerce')

by_category = df.groupby("product_category").agg(
    orders=("order_id", "count"),
    avg_order_usd=("order_total_usd", "mean"),
    return_rate=("returned", "mean"),
    avg_rating=("rating", "mean"),
).round(3).sort_values("avg_order_usd", ascending=False)
by_category
```
